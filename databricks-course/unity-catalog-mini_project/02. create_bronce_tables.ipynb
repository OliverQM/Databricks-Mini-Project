
DROP TABLE IF EXISTS telco_dev.bronce.clientes;
CREATE TABLE IF NOT EXISTS telco_dev.bronce.clientes(
  clienteId INT,
  nombre STRING,
  paterno STRING,
  materno INT,
  documentoIdentidad STRING,
  telefono STRING,
  email STRING,
  direccion STRING
  )
  USING cvs
  OPTIONS(path "abfss://bronze@sadatabricksmini.dfs.core.windows.net/clientes.csv", header "true", delimiter ";")
  

```sql
SELECT * FROM telco_dev.bronce.clientes
```
